```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, sum, mean, min, max

spark = SparkSession.builder \
    .appName("CODTECH Big Data Analysis") \
    .getOrCreate()

print("Spark started successfully")
df = spark.read.csv("dataset.csv", header=True, inferSchema=True)
df.show(5)
df.printSchema()
print(df.columns)
df.describe().show()
df.select(
    [sum(col(c).isNull().cast("int")).alias(c) for c in df.columns]
).show()
df_clean = df.dropna()
print("After removing null values:")
df_clean.show(5)
df_clean.groupBy("passenger_count").count().show()
df_clean.select(
    mean("trip_distance").alias("avg_trip_distance"),
    max("trip_distance").alias("max_trip_distance"),
    min("trip_distance").alias("min_trip_distance")
).show()
df_clean.select(
    mean("fare_amount").alias("avg_fare"),
    max("fare_amount").alias("max_fare"),
    min("fare_amount").alias("min_fare")
).show()
df_clean.groupBy("payment_type").count().show()
print("Big Data Analysis Completed Successfully")
```
